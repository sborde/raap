# Calculate within cell feature means

```python
import numpy as np
import pandas as pd
```

```python
feature_names = ['amplitude', 'half_width']
spike_types = ['AIS-AP', 'UNKNOWN', 'RA-AP']
```

```python
path = 'E:/Data/BogiSejtek/ForManuscript'
file_names = ['B4_d2t1c1', 'Rat121_d1t1c2', 'Rat155_d1t1c3']
```

```python
results = []

for file_name in file_names:
    df = pd.read_csv(f'{path}/{file_name}.csv')
    cell_result = {
        'name': file_name,
        'cell_type': df.iloc[0]['cell_type'],
    }
    for spike_type in spike_types:
        sub_df = df[df['spike_type'] == spike_type]
        cell_result[f'n_{spike_type}'] = len(sub_df)
    for spike_type in spike_types:
        sub_df = df[df['spike_type'] == spike_type]
        cell_result[f'p_{spike_type}'] = len(sub_df) / len(df)
    for spike_type in spike_types:
        sub_df = df[df['spike_type'] == spike_type]
        for feature_name in feature_names:
            cell_result[f'{feature_name}_mean_{spike_type}'] = sub_df[feature_name].mean()
            cell_result[f'{feature_name}_std_{spike_type}'] = sub_df[feature_name].std()
            cell_result[f'{feature_name}_sem_{spike_type}'] = sub_df[feature_name].std() / np.sqrt(len(sub_df))

    results.append(cell_result)
results_df = pd.DataFrame(results)
results_df.to_excel(f'{path}/within_cell_feature_means.xlsx', index=False)
```
