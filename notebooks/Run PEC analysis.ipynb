# Run PEC analysis
The original `run_pec_analysis.py` converted into notebook.

```python
from raap.triggered_firing.io_utils import list_input_files, load_customized_config
from raap.phase_event_coupling.runners import IndividualSpikeTypePECAnalysis
from raap.phase_event_coupling.group_summarizers import run_grouped_analysis

CONFIG_NAME = 'phase_event_coupling'

run_config = load_customized_config(CONFIG_NAME)
analysis = IndividualSpikeTypePECAnalysis(run_config)

input_files = list_input_files(run_config['data'], run_config['input'])
for input_file in input_files:
    try:
        print(f"Processing {input_file.name}...", end=" ")
        analysis(input_file)
        print("done.")
    except Exception as e:
        print(
            f"Some error occurred during processing {input_file.name}: {e}"
        )

if run_config['grouping']['do']:
    run_config['output']['suffix'] = analysis.suffix + '_histogram'
    run_grouped_analysis(run_config)
```
