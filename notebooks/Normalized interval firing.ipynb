# Normalized interval firing histogram

This notebook calculates the AIS-AP and RA-AP firing rate during a normalized interval.

First, the specified episode intervals are collected from the recording, then each interval will be divided to the given number of histogram bins and the number of spikes of both categories will be counted.

## Analysis parameters

Here you can set the histogram parameters.

You can specify the spike types that you want to analyze during the intervals.

```python
params.analysis.spike_types = ['AIS-AP', 'RA-AP']
```

You can specify the episode's name in which you want to analyze the firing rate.

```python
params.analysis.episode_name = 'spdl'
```

 - `values`: unit of the histogram, possible values: `'count'`, `'freq'`, `'prob'`
 - `bin_range`: range to calculate the firing histogram relative to the interval's start (0) and end (1). Negative values refer before, values above 1 refer to after the interval.
 - `bin_size`: size of the normalized bin, relative to the duration of the interval

```python
params.analysis.values = 'prob'
params.analysis.bin_range = (-0.5, 1.5)
params.analysis.bin_size = 0.1
```

You can set the y limit of the plot. The limit of the x axis will be automatically set to the `bin_range`.

```python
params.analysis.ylim = (0, 0.5)
```

You can set the Gaussian kernel of the smoothing or set to 0 if you want raw histogram.

```python
params.analysis.smooth_size = 0
```

You can customize the raster plot.
 - `show_raster`: set `True` if you want to show the raster above the histogram
 - `raster_size`: length of the raster tick
 - `raster_band_color`: color of the band to mark interval with any Ra-Ap spikes

```python
params.analysis.show_raster = True
params.analysis.raster_size = 0.5
params.analysis.raster_band_color = '#bbbbbb'
```

### Cellwise average parameters

 - `interval_condition`: if you leave empty (`[]`), all interals will be used. If you put values in the list, only those intervals will be used to calculate the average where that spike type was present

```python
params.analysis.interval_condition = ['RA-AP']
```
