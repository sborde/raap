# Describe episode intervals per cell

This notebook iterates through the selected cells and calculates and saves the descriptive statistics of the specified episode intervals. It is the same notebook a [Compare firing rate between sleep stages](./Describe%20sleep%20stage%20intervals%20per%20cell.ipynb) but with any type of episode interval.

For each cell, it generates an Excel table with the following properties of each interval:
 - sleep stage name
 - start and end time of the interval
 - interval duration
 - spike count for each spike type

The results can be used in the following analyses:
  - [Compare firing rate between spike types](Compare%20firing%20rate%20between%20spike%20types.ipynb)
